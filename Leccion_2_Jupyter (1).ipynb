# La importancia de las libretas Jupyter en la ciencia de datos

Las libretas Jupyter son una herramienta esencial para el trabajo diario de los científicos de datos. Permiten integrar código, resultados, explicaciones y fórmulas matemáticas en un único documento interactivo. Esta integración no solo facilita el análisis, sino que también establece una forma clara y estructurada de **documentar hallazgos**, permitiendo que otros comprendan, evalúen y repliquen los experimentos realizados. En otras palabras, **las libretas Jupyter son el cuaderno de laboratorio del científico de datos moderno**.

Cada libreta puede verse como un experimento en sí misma, donde se sigue una lógica basada en el **método científico**. Este método parte de una observación inicial, seguida por la formulación de una hipótesis, la ejecución de pruebas o modelos para evaluarla, el análisis de resultados y finalmente la validación o rechazo de dicha hipótesis. En ciencia de datos, esto se traduce en:

1. **Observación**: Exploración de los datos disponibles.
2. **Hipótesis**: Supuestos sobre patrones, relaciones o tendencias.
3. **Experimentación**: Aplicación de técnicas estadísticas o modelos de aprendizaje de máquina (machine learning).
4. **Análisis**: Evaluación de métricas, visualización de resultados.
5. **Conclusión**: Interpretación de hallazgos y toma de decisiones basadas en datos.

Jupyter permite que cada paso de este proceso quede claramente registrado. Usando **Markdown**, los científicos de datos pueden escribir descripciones, justificar decisiones o explicar resultados. Con **Python**, ejecutan código que realiza análisis, limpia datos, entrena modelos o genera gráficos. Y con **LaTeX**, incorporan expresiones matemáticas formales para reforzar la rigurosidad del trabajo.

Este enfoque está profundamente influenciado por la filosofía de **Literate Programming** de **Donald Knuth**, quien propuso que el código no solo debía ser funcional, sino también legible y explicativo para humanos. En ese espíritu, las libretas Jupyter no solo contienen instrucciones para la máquina, sino también una narrativa para quien quiera entender el proceso detrás de los resultados.

En un entorno donde la reproducibilidad, la transparencia y la colaboración son pilares de la ciencia, las libretas Jupyter son mucho más que una herramienta de programación: **son el medio principal con el que los científicos de datos piensan, trabajan y comunican conocimiento**.


## ¿Qué es Markdown?

**Markdown** es un lenguaje de marcado ligero que permite escribir texto con formato de manera sencilla y legible. Se utiliza ampliamente en libretas Jupyter para documentar análisis, explicar código, presentar resultados, escribir fórmulas matemáticas, e incluso insertar imágenes o enlaces.

A diferencia de HTML o LaTeX, Markdown tiene una sintaxis muy simple e intuitiva. Cuando una celda de tipo *Markdown* se ejecuta (`Shift + Enter`), el texto se convierte en contenido formateado.

### Ejemplos comunes de Markdown:

- **Encabezados**:
  - `# Título principal`
  - `## Subtítulo`
  - `### Tercer nivel`

- **Texto en negritas o cursivas**:
  - `**negritas**` → **negritas**
  - `*cursivas*` → *cursivas*

- **Listas**:
  - `- Elemento`
  - `1. Elemento numerado`

- **Enlaces**:
  - `[Texto del enlace](https://ejemplo.com)`

- **Imágenes**:
  - `![Texto alternativo](ruta/imagen.png)`

- **Código en línea**:
  - `` `print("Hola mundo")` `` → `print("Hola mundo")`

- **Bloques de código**:
  \`\`\`python  
  print("Hola Jupyter")  
  \`\`\`

- **Fórmulas matemáticas** (usando LaTeX):
  - `$E = mc^2$` → \( E = mc^2 \)
  - Bloques:  
    ```
    $$
    \sum_{i=1}^n x_i
    $$
    ```

Markdown permite crear libretas bien estructuradas y comprensibles, lo que es clave en ciencia de datos para comunicar análisis de forma clara y profesional.

# ¿Qué es una celda en Jupyter?

Una **celda** es la unidad básica de contenido en una libreta Jupyter. Cada libreta está compuesta por múltiples celdas que se ejecutan de manera independiente, y que pueden contener distintos tipos de contenido.

Existen principalmente dos tipos de celdas:

- **Celda de código (`Code`)**:
  Estas celdas permiten escribir y ejecutar código, generalmente en Python (también se soporta R). Al ejecutarse, el código se procesa por el *kernel* y se muestra el resultado directamente debajo de la celda. Aquí es donde se llevan a cabo cálculos, análisis, creación de gráficos, entrenamiento de modelos, etc.

- **Celda de texto (`Markdown`)**:
  Estas celdas están escritas en el lenguaje de marcado *Markdown*. Se utilizan para documentar, describir procesos, escribir hipótesis, interpretar resultados y también pueden incluir imágenes, listas, enlaces y fórmulas matemáticas utilizando LaTeX.

Este sistema de celdas facilita combinar explicación y ejecución en un mismo documento, promoviendo un enfoque claro, reproducible y didáctico para el análisis de datos.


## ¿Cómo agregar celdas de código y texto en Jupyter?

En Jupyter, puedes agregar nuevas celdas para escribir **código** o **texto (Markdown)** según lo que necesites documentar o ejecutar.

### Para agregar una celda nueva:

- Haz clic en el menú **Insertar /Insert** y elige:
  - **Insertar celda Code** .
  - **Insertar celda Text** l.

También puedes usar atajos del Toolbar:

- Seleccioa **`+Code`** para celda de Código
- Seleccioa **`+Text`** para celda de Markdown

### Para cambiar el tipo de celda:

- Selecciona la celda y en la barra de herramientas usa el menú desplegable que dice **Code** o **Markdown**.
- O usa los siguientes atajos (estando en modo comando):
  - **`Y`** convierte la celda en **Code** (código).
  - **`M`** convierte la celda en **Markdown** (texto explicativo).

### Recuerda:

- Las **celdas de código** se ejecutan con `Shift + Enter`, mostrando el resultado debajo.
- Las **celdas de Markdown** también se ejecutan con `Shift + Enter`, pero su contenido se renderiza como texto formateado.

Usar adecuadamente ambos tipos de celdas permite combinar código funcional con documentación clara y comprensible, ideal para comunicar tus ideas en ciencia de datos.


## ¿Qué es un Kernel en Jupyter?

Un **kernel** en Jupyter es el motor que ejecuta el código que escribes en las celdas de tipo código. Cada kernel corresponde a un lenguaje de programación específico y mantiene en memoria el estado de las variables, funciones y resultados durante toda la sesión de trabajo.

Cuando ejecutas una celda, Jupyter envía el contenido al kernel correspondiente, que procesa el código y devuelve la salida que se muestra debajo de la celda.

### Ejemplos de kernels utilizados en ciencia de datos:

- **IPython (Python)**: El más común. Permite ejecutar código Python y utilizar bibliotecas como NumPy, pandas, matplotlib, scikit-learn, TensorFlow, entre otras.
- **IRkernel (R)**: Kernel para el lenguaje R, muy utilizado en estadística y análisis de datos.
- **Julia**: Kernel para el lenguaje Julia, diseñado para alto rendimiento en cálculos numéricos y ciencia computacional.
- **Apache Toree**: Kernel que permite ejecutar código Scala, Java, SQL y otros lenguajes en un entorno Apache Spark.
- **Sos (Script of Scripts)**: Kernel que permite trabajar con múltiples lenguajes (Python, R, Bash, etc.) en una misma libreta.

El kernel puede ser reiniciado, interrumpido o cambiado desde el menú de Jupyter. Esta flexibilidad es clave en el trabajo exploratorio de la ciencia de datos, donde se necesita probar y modificar fragmentos de código de forma interactiva.

## Escribir ecuaciones matemáticas en Markdown

$ y = \beta_{0} + 2\beta_1^2 $

# LaTeX

## ¿Qué es LaTeX?

**LaTeX** (pronunciado *lá-tej* o *léi-tec*) es un sistema de composición de textos especialmente diseñado para la creación de documentos científicos, matemáticos y técnicos de alta calidad tipográfica.

A diferencia de los procesadores de texto tradicionales (como Word), LaTeX no se basa en lo que ves es lo que obtienes (WYSIWYG), sino en el uso de **comandos de marcado** para estructurar y formatear el contenido. Esto le da al autor un **control total sobre el diseño**, ideal para fórmulas matemáticas, referencias cruzadas, bibliografías, tablas complejas y más.

---

## 📜 Breve historia de LaTeX

### Origen

- En la década de **1970**, el informático **Donald E. Knuth** desarrolló un sistema tipográfico llamado **TeX**, motivado por la mala calidad de impresión de las fórmulas matemáticas en sus propios libros.
- TeX se centró en la precisión tipográfica y el control detallado sobre el diseño de páginas.

### Nacimiento de LaTeX

- En **1984**, **Leslie Lamport** creó **LaTeX** como un conjunto de macros sobre TeX. Su objetivo fue hacer más **fácil y accesible** el uso de TeX para científicos y académicos, con una sintaxis más amigable.

### Evolución y uso actual

- Desde entonces, LaTeX se ha convertido en el estándar de facto en campos como:
  - Matemáticas
  - Física
  - Ciencias de la computación
  - Ingeniería
  - Economía y estadística
- Es ampliamente usado para artículos científicos, tesis, libros, presentaciones académicas e incluso para escribir partituras musicales o textos filosóficos complejos.

---

## ✍️ ¿Por qué usar LaTeX hoy?

- Permite escribir **fórmulas matemáticas complejas** con facilidad.
- Produce documentos con **calidad editorial profesional**.
- Maneja bien bibliografías, tablas, imágenes y referencias cruzadas.
- Es **gratuito y de código abierto**.
- En entornos como **Jupyter Notebooks**, puedes usar LaTeX para escribir ecuaciones dentro de celdas Markdown.

---

> 🧠 En resumen: **LaTeX** no es solo una herramienta de escritura, sino una forma de pensar en la estructura del conocimiento, con precisión, claridad y elegancia.

## 🧠 LaTeX en acción: redes neuronales y backpropagation

Ahora que ya sabes cómo usar LaTeX para escribir ecuaciones, veamos cómo aplicarlo en un contexto real de ciencia de datos: las **redes neuronales** y su entrenamiento mediante **backpropagation**.

---

## 🔁 ¿Qué es una red neuronal?

Una **red neuronal artificial** es una función compuesta de múltiples capas de funciones lineales seguidas por funciones de activación no lineales. La salida de una capa es la entrada de la siguiente.

### Ecuación general de una capa:

```latex
$$
\mathbf{z}^{(l)} = \mathbf{W}^{(l)} \mathbf{a}^{(l-1)} + \mathbf{b}^{(l)}
$$

$$
\mathbf{a}^{(l)} = f\left(\mathbf{z}^{(l)}\right)
$$
````

Donde:

* $\mathbf{z}^{(l)}$ es el vector de activaciones lineales en la capa $l$
* $\mathbf{W}^{(l)}$ es la matriz de pesos
* $\mathbf{b}^{(l)}$ es el vector de sesgo
* $\mathbf{a}^{(l-1)}$ es la salida de la capa anterior
* $f$ es la función de activación (por ejemplo, sigmoide, ReLU)

---

## 🎯 Función de pérdida

El objetivo del entrenamiento es minimizar una **función de pérdida** $\mathcal{L}$ que mide la diferencia entre las predicciones y las etiquetas verdaderas:

```latex
$$
\mathcal{L} = \frac{1}{n} \sum_{i=1}^n \left( \hat{y}_i - y_i \right)^2
$$
```

(Para el caso de regresión con error cuadrático medio)

---

## 🔄 ¿Qué es backpropagation?

**Backpropagation** (retropropagación) es el algoritmo que se usa para **ajustar los pesos** de la red utilizando el gradiente de la función de pérdida con respecto a cada parámetro.

La idea central es aplicar la **regla de la cadena** para propagar los errores desde la capa de salida hacia atrás, actualizando los pesos en el camino.

---

## 🔽 Derivadas clave en backpropagation

### Derivada del error respecto a la salida de la última capa:

```latex
$$
\delta^{(L)} = \nabla_a \mathcal{L} \circ f'\left(\mathbf{z}^{(L)}\right)
$$
```

### Propagación del error hacia capas anteriores:

```latex
$$
\delta^{(l)} = \left( \mathbf{W}^{(l+1)T} \delta^{(l+1)} \right) \circ f'\left(\mathbf{z}^{(l)}\right)
$$
```

### Gradientes de los parámetros:

```latex
$$
\frac{\partial \mathcal{L}}{\partial \mathbf{W}^{(l)}} = \delta^{(l)} \cdot \left(\mathbf{a}^{(l-1)}\right)^T
$$

$$
\frac{\partial \mathcal{L}}{\partial \mathbf{b}^{(l)}} = \delta^{(l)}
$$
```

---

## 📉 Actualización de parámetros

Una vez calculados los gradientes, actualizamos los pesos con descenso del gradiente:

```latex
$$
\mathbf{W}^{(l)} := \mathbf{W}^{(l)} - \eta \cdot \frac{\partial \mathcal{L}}{\partial \mathbf{W}^{(l)}}
$$

$$
\mathbf{b}^{(l)} := \mathbf{b}^{(l)} - \eta \cdot \frac{\partial \mathcal{L}}{\partial \mathbf{b}^{(l)}}
$$
```

Donde $\eta$ es la **tasa de aprendizaje**.

# Comandos de tipo Magic

Las celdas de código en Jupyter tienen una serie de "comandos mágicos" (*magic commands*) que te permiten ejecutar comandos especiales que solo funcionan dentro de una libreta.  
Un ejemplo es ver el historial de comandos del notebook o ejecutar otros lenguajes de programación distintos al del kernel actual.

Los comandos mágicos se escriben anteponiendo el símbolo `%` (y para ejecutar bloques en otros lenguajes, `%%`).



## Comandos mágicos comunes en Jupyter

Los **comandos mágicos** en Jupyter permiten extender la funcionalidad del entorno con comandos especiales que no son parte del lenguaje Python, pero que son útiles para tareas comunes como medir tiempos, cargar archivos, ejecutar scripts, o interactuar con el sistema operativo.

Se dividen en:

- **Comandos mágicos de línea**: comienzan con `%` y afectan solo una línea.
- **Comandos mágicos de celda**: comienzan con `%%` y afectan toda la celda.

---

### 🔹 Comandos mágicos de línea (`%`)

- `%time`: mide el tiempo de ejecución de una línea de código.
  ```python
  %time sum(range(1000000))
````

* `%timeit`: ejecuta la línea varias veces y da un promedio de tiempo.

  ```python
  %timeit sum(range(1000000))
  ```

* `%who`: lista las variables definidas en el espacio de trabajo.

* `%whos`: lista las variables con más detalle (tipo, valor, etc.).

* `%pwd`: muestra el directorio de trabajo actual.

* `%cd`: cambia el directorio de trabajo.

* `%history`: muestra el historial de comandos ejecutados.

* `%run script.py`: ejecuta un archivo de Python externo.

* `%matplotlib inline`: permite mostrar gráficos de Matplotlib directamente en la libreta.

---

### 🔸 Comandos mágicos de celda (`%%`)

* `%%time`: mide el tiempo de ejecución de toda la celda.

* `%%timeit`: realiza múltiples ejecuciones de toda la celda y da una media de tiempo.

* `%%writefile archivo.py`: guarda el contenido de la celda como un archivo.

  ```python
  %%writefile ejemplo.py
  print("Este código será guardado en un archivo")
  ```

* `%%bash`: permite ejecutar comandos de bash directamente en la celda.

  ```bash
  %%bash
  echo "Hola desde bash"
  ```

* `%%html`, `%%latex`, `%%javascript`: permiten renderizar directamente contenido en esos lenguajes.

---

> 🧪 Estos comandos mágicos son exclusivos de los notebooks Jupyter y te permiten combinar análisis, medición de rendimiento, depuración y automatización de tareas de forma muy eficiente.

# Mejores prácticas para escribir un notebook de Jupyter reproducible

Un buen notebook de Jupyter no solo ejecuta código: **cuenta la historia de un experimento**, documenta la hipótesis, describe los pasos del análisis y permite que otros (o tú mismo en el futuro) puedan **repetir el experimento y validar los resultados**.

Aquí te compartimos las mejores prácticas para lograrlo:

### 1. Explica la hipótesis al inicio
Utiliza una celda Markdown para **formular claramente la hipótesis** o pregunta que deseas explorar. Describe el contexto, los objetivos del análisis y lo que esperas encontrar.

---

### 2. Organiza el notebook en secciones
Usa encabezados Markdown (`#`, `##`, `###`) para estructurar el notebook en secciones como:

- Importación de librerías
- Carga de datos
- Limpieza y preparación
- Análisis exploratorio
- Modelado
- Evaluación de resultados
- Conclusiones

Esto facilita la lectura y navegación.

---

### 3. Ejecuta las celdas en orden
Mantén un flujo lógico **de arriba hacia abajo**. Asegúrate de que el notebook pueda ejecutarse de principio a fin usando `Kernel > Restart & Run All` sin errores.

---

### 4. Documenta con Markdown y comentarios
Usa celdas Markdown para explicar qué estás haciendo y por qué. Añade comentarios en el código para describir su propósito. Esto hace el análisis comprensible incluso para quienes no escribieron el notebook.

---

### 5. Usa nombres de variables descriptivos
Evita nombres ambiguos como `x` o `df2`. Prefiere nombres como `ventas_mensuales` o `modelo_arbol`. Esto mejora la legibilidad y mantenimiento.

---

### 6. Controla fuentes de aleatoriedad
Si usas funciones aleatorias (por ejemplo, para dividir datos o entrenar modelos), **fija una semilla aleatoria** (`random_state`) para asegurar que los resultados sean reproducibles.

---

### 7. Evita dejar resultados "caché"
Antes de compartir el notebook, **reinicia el kernel y ejecuta todo desde cero**. Así te aseguras de que no hay resultados "fantasma" de ejecuciones anteriores.

---

### 8. Guarda versiones del dataset o documenta su origen
Si el dataset cambia con el tiempo, guarda una copia local o indica claramente la fuente, la fecha de descarga y los pasos para obtenerlo. Así otros podrán acceder exactamente al mismo conjunto de datos.

---

### 9. Advierte sobre pasos costosos
Si alguna celda tarda mucho (como entrenamiento de modelos complejos), indícalo con una celda Markdown para que el lector sepa qué esperar.

---

### 10. Finaliza con conclusiones claras
Resume tus hallazgos en una celda Markdown al final. Compara los resultados con la hipótesis inicial y comenta si fue confirmada o refutada, y qué pasos seguirías después.

---

> 🧪 Siguiendo estas prácticas, tu notebook se convierte en un documento científico completo, claro y reproducible, capaz de comunicar conocimiento y facilitar el aprendizaje y la colaboración.

# 🧪 Ejercicio: Resolución de una ecuación de tercer grado con Newton-Raphson




## 🧠 Ejercicio: Resolver una ecuación cúbica con el método de Newton-Raphson

En este ejercicio resolverás la siguiente ecuación cúbica usando el **método de Newton-Raphson**:

$$
f(x) = x^3 - 6x^2 + 11x - 6 = 0
$$

Esta ecuación tiene raíces reales, y tu tarea será encontrar una de ellas utilizando el método iterativo de Newton-Raphson.

---

### ✍️ Tareas:

1. **Escribe una celda Markdown** con:
   - La función \( f(x) \) y su derivada \( f'(x) \), escritas en LaTeX.
   - Una breve explicación del método de Newton-Raphson.

2. **Escribe una celda de código Python** que:
   - Implemente el método de Newton-Raphson.
   - Use una estimación inicial \( x_0 = 3.5 \).
   - Itere hasta que el error sea menor a \( 10^{-6} \).
   - Muestre el resultado y el número de iteraciones.

3. **Documenta tu código** con comentarios explicativos.

---

### 📌 Fórmulas:

- Iteración de Newton-Raphson:

$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$

- Derivada de \( f(x) \):

$$
f'(x) = 3x^2 - 12x + 11
$$

### 💻 Ejemplo de código base sugerido (que el alumno puede mejorar):

```python
def f(x):
    return x**3 - 6*x**2 + 11*x - 6

def df(x):
    return 3*x**2 - 12*x + 11

def newton_raphson(x0, tol=1e-6, max_iter=100):
    x = x0
    for i in range(max_iter):
        fx = f(x)
        dfx = df(x)
        if abs(dfx) < 1e-10:
            print("La derivada es muy pequeña.")
            return None
        x_new = x - fx/dfx
        if abs(x_new - x) < tol:
            return x_new, i+1
        x = x_new
    print("No convergió en el número de iteraciones dado.")
    return None

raiz, iteraciones = newton_raphson(3.5)
print(f"Raíz encontrada: {raiz} en {iteraciones} iteraciones")
```
